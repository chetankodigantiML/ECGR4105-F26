# Problem 1

```python
import numpy as np
import pandas as pd
import random as ran
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/drive')
```

```python
file_Path = '/content/drive/My Drive/Fall26/ECGR4105/HW1/Dataset/D3.csv'
data = pd.read_csv(file_Path)
```

```python
x1 = (data.values[:,0])
x1 = x1.reshape(len(x1),1)

x2 = data.values[:,1]
x2 = x2.reshape(len(x2),1)

x3 = data.values[:,2]
x3 = x3.reshape(len(x3),1)

y = data.values[:,3]
```

```python
X1 = np.hstack((np.ones((len(x1),1)),x1))
X2 = np.hstack((np.ones((len(x2),1)),x2))
X3 = np.hstack((np.ones((len(x3),1)),x3))
```

```python
def lin_reg(x, y, alpha, iterations):
  m = len(x)
  w = [0]*len(x[0])
  cost = np.zeros(iterations)
  i = 0
  for i in range(iterations):
    pred = x.dot(w)
    errors = np.subtract(pred, y)
    w -= (alpha / m) * x.transpose().dot(errors)
    cost[i] = 1/(2 * m) * np.sum(np.square(errors))
  return w, cost
```

```python
X1_weights, X1_cost = lin_reg(X1, y, 0.1, 400)
plt.scatter(X1[:,1], y, marker='+')
plt.rcParams["figure.figsize"] = (10, 10)
plt.xlabel('X1 values')
plt.ylabel('Y values')
plt.scatter(X1[: ,1], X1.dot(X1_weights))
plt.title("Y vs X1 Linear Regression Plot")
plt.grid(True)
plt.show()
```

```python
plt.plot(X1_cost)
plt.title("X1 Cost Plot")
plt.xlabel('Iteration #')
plt.ylabel('X1 Cost Plot')
plt.grid(True)
plt.show()
```

```python
X2_weights, X2_cost = lin_reg(X2, y, 0.01, 110)
plt.scatter(X2[:,1], y)
plt.scatter(X2[: ,1], X2.dot(X2_weights))
plt.rcParams["figure.figsize"] = (10, 10)
plt.xlabel('X2 values')
plt.ylabel('Y values')
plt.title("Y vs X2 Linear Regression Plot")
plt.grid(True)
plt.show()
```

```python
plt.plot(X2_cost)
plt.title("X2 Cost Plot")
plt.xlabel('Iteration #')
plt.ylabel('Cost Value')
plt.grid(True)
plt.show()
```

```python
X3_weights, X3_cost = lin_reg(X3, y, 0.1, 300)
plt.scatter(X3[:,1], y)
plt.scatter(X3[: ,1], X3.dot(X3_weights))
plt.title("Y vs X3 Linear Regression Plot")
plt.xlabel('X3 values')
plt.ylabel('Y values')
plt.grid(True)
plt.show()
```

```python
plt.plot(X3_cost)
plt.title("X3 Cost Plot")
plt.xlabel('Iteration #')
plt.ylabel('Cost Value')
plt.grid(True)
plt.show()
```
